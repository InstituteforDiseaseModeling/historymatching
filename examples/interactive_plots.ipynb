# Testing interactive plots

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import history_matching as hm

%matplotlib inline
```

```python
# Generate test data
m = 3  # predictors or independent data
n = 20 # number of data points/rows
p = 5  # dependent data


def generate_data(m, n, p):
# Generate random independent data for x
    x = pd.DataFrame(np.random.rand(n, m), columns=[f'input_{i+1}' for i in range(m)])

    # Simple function for y
    y_data = []
    for j in range(p):
        y_column = x.iloc[:, :min(m, j+1)].sum(axis=1)*(j+1)
        y_data.append(y_column)

    # Create y DataFrame
    y = pd.DataFrame(np.array(y_data).T, columns=[f'output_{j+1}' for j in range(p)])

    return x, y

# Let's generate the data
x, y = generate_data(m, n, p)

print("x DataFrame:")
print(x.head())
print("\ny DataFrame:")
print(y.head())
```

```python
diagnostics = hm.features.Diagnostics(x, y)
```
